```python
import pandas as pd
fraud_data = pd.read_csv("Fraud_Data.csv")
ip_country = pd.read_csv("IpAddress_to_Country.csv")
fraud_data_sorted = fraud_data.sort_values('ip_address')
ip_country_sorted = ip_country.sort_values('lower_bound_ip_address')

merged_data = pd.merge_asof(
    fraud_data_sorted,
    ip_country_sorted,
    left_on="ip_address",
    right_on="lower_bound_ip_address",
    direction="backward"
)

merged_data = merged_data[merged_data["ip_address"] <= merged_data["upper_bound_ip_address"]]

merged_data.to_csv('final_data.csv', index=False)
```
